# Acquiring the dataset

The initial seed dataset comes from [arXiv Scientific Research Papers Dataset](https://www.kaggle.com/datasets/sumitm004/arxiv-scientific-research-papers-dataset).

```python
import kagglehub

df = kagglehub.dataset_load(
    adapter=kagglehub.KaggleDatasetAdapter.POLARS,
    handle="sumitm004/arxiv-scientific-research-papers-dataset",
    path="arXiv_scientific_dataset.csv"
)

df.head().collect()
```
