# Calculating embeddings from the data

The following notebook showcases embedding of the over 200k summaries in the arXiv dataset. This might take long and you might want to adjust the batch size!

Colab with a T4 GPU can do this in about 18 minutes. Don't use a TPU, it's slower even than my local CPU.

```python
import kagglehub
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
df = kagglehub.dataset_load(
    adapter=kagglehub.KaggleDatasetAdapter.POLARS,
    handle="sumitm004/arxiv-scientific-research-papers-dataset",
    path="arXiv_scientific_dataset.csv"
)

df.head().collect()
```

```python
summaries = df.select("summary").collect()["summary"].to_list()

print(len(summaries))

embeddings = model.encode(summaries, batch_size=10,
        show_progress_bar=True,
        convert_to_numpy=True,)
```
